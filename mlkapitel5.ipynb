## Faktafrågor

### Uppgift 1. 
*Curse of dimensionality* (dimensionsförbannelsen) innebär att när antalet variabler/dimensioner i ett dataset ökar blir problmet för maskininlärningsmodellen betydligt mer komplext.
När dimensionerna blir många:
- krävs mer data för att modellen ska kunna lära sig bra,
- blir datapunkterna glesare,
- ökar risken för överanspassning (overfitting),
- kan beräkningar och modellträning ta längre tid,
- blir det svårare att visualisera och tolka datan

Ett exempel är att 100 observationer kan vara ganska mycket information om man har två variabler, men väldigt lite information om man har flera hundra variabler

### Uppgift 2.
Dimensionsreducering innebär att man minskar antalet variabler/dimensioner i ett dataset, samtidigt som man försöker behålla så mycket viktig information som möjligt.

Det görs bland annat för att:
- minska risken för overfitting,
- göra modeller snabbare,
- minska mängden data som behöver bearbetas,
- ta bort deundanta eller mindre viktiga egenskaper,
- göra data lättare att visualiser,
- ibland förbättra modellens prestanda.
Ett vanligt sätt att göra dimensionsreducering är **PCA (Principal Component Analysis)**

### Uppgift 3. 
PCA (Principal Component Analysis) fungerar i korthet ut på att hitta de riktningar i datan där spridningen (variansen) är som störst, och använda dessa för att komprimera ner dimensionerna.

Med utgångspunkt i figuren kan man förklara processen i tre steg:

1. **Hitta huvudkomponenter (vänstra diagrammet):**
Datan i det vänstra diagrammet har två ursprungliga dimensioner ($X_1$ och $X_2$). Istället för att bara titta på axlarna letar PCA efter linjer som fångar upp mest information. Den första huvudkomponenten ($C_1$, den helt dragna linjen) dras längs med den riktning där datan är mest utspridd. Den andra komponenten ($C_2$, den streckade linjen) står vinkelrätt mot den första och fångar upp återstående varians.

2. **Projicera ner på en ny axel (högra diagrammet):**
När vi vill minska dimensionerna (t.ex. från 2D till 1D) projicerar vi datapunkterna vinkelrätt ner på en av dessa linjer.

- Om vi projicerar på $C_1$ (översta grafen till höger, $Z_1$) behålls spridningen i datan väldigt väl eftersom linjen följer molnets längsta utsträckning.
- Om vi istället skulle projicera på en dålig axel (som de två undre graferna visar) pressas punkterna ihop och mycket mer information (varians) går förlorad.

3. **Resultat:**
Genom att välja den axel som har störst spridning ($C_1$) maximerar vi den information som sparas, vilket gör att vi kan reducera antalet dimensioner (från 2 till 1) med minimal informationsförlust.

### Uppgift 4. 
Kernel OCA kan utvärderas genom att udersöka hur bra den reducerade representationen fungerar för den uppgift man vill lösa.
Man kan exempelvis:
- jämföra modellens prestanda före och efter dimensionsreduceringen,
- använda de reducerade komponenterna som input till en maskininlärningsmodell och jämföra exempelvis **accuracy, precision, recall eller RMSE**, 
- undersöka hur mycket information som verkar bevaras,
- visualisera datan efter reduktionen och se om olika klasser blir tydligare separerade.

Det finns alltså inte nödvändigtvis ett enda mått som säger att kernel PCA är "bra". Man bör istället utvärdera om den nya representationen faktiskt hjälper den efterföljande ML-uppgiften.

### Uppgift 6. 
Efter PCA blir tolkningen av de ursprungliga variablerna svårare.

Det beror på att PCA skapar nya variabler, alltså principal components, som är kombinationer av de ursprungliga variablerna.

Exempelvis kan en principal component bestå av:

PC1 = 0,6 × längd + 0,5 × vikt − 0,2 × ålder

PC1 är alltså inte längre en variabel som har en direkt och enkel betydelse som exempelvis "vikt" eller "ålder".

Fördelarna är att vi kan reducera antalet dimensioner och behålla mycket av variationen i datan. Nackdelen är att tolkbarheten minskar, eftersom de nya komponenterna representerar kombinationer av flera ursprungliga variabler.

Därför är PCA ofta användbart när modellens prestanda och dimensionsreducering är viktigare än att kunna tolka varje enskild variabel.

### Uppgift 8. 

Koden demonstrerar dimensionalitetsreduktion med hjälp av PCA (**Principal Component Analysis**), vilket är en populär teknik inom oövervakad maskininlärning för att banta ner antalet variabler (features) i ett dataset samtidigt som man behåller så mycket som möjligt av informationen (variansen).

Här är en uppdelning rad för rad av vad koden gör:
1. **Skapar ett slumpmässigt dataset:**
`X = np.random.rand(1000, 3)` skapar en matris med 1 000 rader (observationer) och 3 kolumner (features) fyllda med slumpmässiga värden mellan 0 och 1. `print(X[0:5])` visar de första 5 raderna.

2. **Reducerar dimensionerna:**
`pca = PCA(n_components=2)` förbereder en PCA-modell som är inställd på att komprimera datan från 3 dimensioner ner till 2.
`X2D = pca.fit_transform(X)` anpassar modellen till datan och omvandlar de ursprungliga 3 kolumnerna till 2 nya dimensioner (eller "huvudkomponenter"). `print(X2D[0:5])` visar hur de första 5 raderna ser ut efter nedskalningen.

3. **Återskapar datan (Inverse transform):**
`X3D_inv = pca.inverse_transform(X2D)` försöker "backa" operationen och räkna om de 2 dimensionerna till 3 dimensioner igen.

4. **Jämför originalet med den återskapade datan:**
`np.allclose(X3D_inv, X)` kollar om den återskapade matrisen är i princip identisk med originalmatrisen. Svaret här blir False, eftersom komprimeringen från 3 till 2 dimensioner innebär att en del information (varians) oundvikligen har gått förlorad på vägen.

  - **Linjära modeller (t.ex. Linear Regression):** Påverkas knappt alls i träffsäkerhet eftersom linjära samband bevaras väl, men beräkningarna blir mer komprimerade.


#### Python-kod för att köra PCA på datasetet

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# 1. Ladda datasetet
df = pd.read_csv("car_price_dataset.csv", sep=';')

X = df.drop(columns=['Price'])
y = df['Price']

# Identifiera numeriska och kategoriska kolumner
numeric_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = X.select_dtypes(include=['object']).columns.tolist()
```

```python
# 2. Dela upp datan i Train, Validation och Test (70% / 15% / 15%)
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42
)
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, test_size=0.1765, random_state=42
)

print(f"Träningsset:    {X_train.shape[0]} rader")
print(f"Valideringsset: {X_val.shape[0]} rader")
print(f"Testset:        {X_test.shape[0]} rader\n")
```

```python
# 3. Preprocessors (standard för baslinje, tät för PCA)
preprocessor_standard = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]
)

preprocessor_dense = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_cols)
    ]
)
```

```python
# --- 4. MODELL UTAN PCA (Baslinje) ---
baseline_pipeline = Pipeline([
    ('preprocessor', preprocessor_standard),
    ('model', RandomForestRegressor(random_state=42, n_estimators=50))
])

baseline_pipeline.fit(X_train, y_train)
val_preds_base = baseline_pipeline.predict(X_val)

r2_base = r2_score(y_val, val_preds_base)
rmse_base = np.sqrt(mean_squared_error(y_val, val_preds_base))
```

```python
# --- 5. MODELL MED PCA ---
pca_pipeline = Pipeline([
    ('preprocessor', preprocessor_dense),
    ('pca', PCA(n_components=0.95, random_state=42)),
    ('model', RandomForestRegressor(random_state=42, n_estimators=50))
])

pca_pipeline.fit(X_train, y_train)
val_preds_pca = pca_pipeline.predict(X_val)

r2_pca = r2_score(y_val, val_preds_pca)
rmse_pca = np.sqrt(mean_squared_error(y_val, val_preds_pca))
n_components = pca_pipeline.named_steps['pca'].n_components_
```

```python
# --- 6. UTVÄRDERING OCH JÄMFÖRELSE PÅ VAL-SET ---
print("========================================")
print("       MODELLJÄMFÖRELSE (BILPRISER)     ")
print("========================================")
print(f"Utan PCA -> R2: {r2_base:.4f} | RMSE: {rmse_base:.2f}")
print(f"Med PCA  -> R2: {r2_pca:.4f} | RMSE: {rmse_pca:.2f} (Komponenter: {n_components})")
```
