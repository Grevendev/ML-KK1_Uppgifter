### Uppgift 5. Bias-variance trade-off handlar om balansen mellan att en modell är för enkel och att den är för känslig för träningsdatan.
- **Bias** 
**Hög bias** betyder att modellen är för enkel och gör för starka antaganden. -> Modellen underfittar. Modellen kan alltså inte fånga de riktiga sambanden.
- **Variance**
**Hög Variance** betyder att modellen är väldigt känslig för exakt vilken träningsdatan den får. Modellen **överanpassar (overfittar)**.
- **Trade-off**
Vi vill hitta en balans
````
För enkel                 Lagom                  För komplex
    │                       │                        │
    ▼                       ▼                        ▼
Hög bias               Bra generalisering      Hög variance
Underfitting                                      Overfitting
````

Varför är mer komplexa modeller inte alltid bättre?
En mer komplex modell har större möjlighet att beskriva träningsdatan väldigt exakt. Men problemet är att den också börja lära sig brus och slumpmässiga detaljer. 

````
Enkel modell:
"Större hus → högre pris"

Komplex modell:
"Större hus → högre pris, men exakt det här området,
den här lilla avvikelsen och den här enskilda observationen
påverkar också..."
````

Den komplexa modellen kanske får lägre fel på träningsdata, men högre fel på testdata. 

### Uppgift 6. 
a) **Linjär regression**
Linjär regression försöker hitta en rak linje som beskriver sambandet mellan inputvariabkerna och det värde vi vill förustäga.
````
Exempel: 
Husets storlek -> husets pris
````
Modellen försöker hitta de parametrarna som gör att skillnaden mellan de verkliga och förutsagda värdena blir så liten som möjligt

b) **Ridge Regression**
Ridge är egentligen **linjär regression + regularisering.**
Den lägger till ett straff för att modellens koefficienter blir väldigt stora. Det gör att modellen inte lika lätt överanpassat träningsdatan.

c) **Lassa Regression**
Lasso är också linjär regression med regularisering, men använder en annan typ av straff.
En viktig egenskap är att Lassa kan göra vissa koefficienter till exakt 0.
Den innebär att Lasso automatiskt kan välja bort vissa variabler.

d) **Elastic Net**
Elastic Net kombinerar Ridge och Lasso.
Den använder allstå både typerna av regularisering.
Det gör att den kan:
- begränsa stora koefficienter, som Ridge
- sätta vissa koefficienter till 0, som Lasso

e) **Support Vector Machine (SVM)**
En Support Vector Machine försöker hitta en funktion som passar datan samtidigt som modellen försöker hålla en så bred marginal som möjligt runt sina prediktioner.

För regression kallas varianten Support Vector Regression (SVR).

Med hjälp av olika kernels kan modellen även hantera icke-linjära samband.

f) **Beslutsträd**
Ett beslutsträd delar upp datan genom en serie frågor.
Exempel: 
````
Är inkomsten > 50 000?
       /       \
     Ja         Nej
     /           \
Är huset stort?   ...
````
Till slut hamnar observationen i ett **leaf**, där modellen ger en prediktion.
För regression kan exempelvis medelvärdet av träningsobservationerna i ett leaf användas som prediktion. 

g) **Ensemble learning**
Ensemble learning innebär att man kombinerar flera modeller för att skapa en bättre modell.
Tanken är ungär:
````
"Flear modeller tillsammans kan ge en bättre prediktion än en enskild modell".
````
Exempel är:
- Random Forest
- Gradient Boosting
- AdaBoost
Det finns olika sätt att kombinera modeller, exempelvis genom att medelvärdesbilda deras prediktioner.

h) **Random Forest**
Random Forest är en typ av ensemble learning som bestpr av många beslutsträd.
Varje träd tränas på en variation av datan och/eller feature. När vi ska göra en regression får varje trä göra en prediktion och Random Forest kombinerar sedan dessa, vanligtvis genom att ta medelvärdet. 
Exempel:
````
Träd 1 → 350 000 kr
Träd 2 → 370 000 kr
Träd 3 → 360 000 kr
Träd 4 → 340 000 kr
       ↓
Random Forest
       ↓
≈ 355 000 kr
````
Det gör modellen betydligt mer robust än ett enskilt beslutsträd.

### Uppgift 7. Skillnaden handlar framför allt om hur lätt det är att förstå varför modellen ger en viss prediktion.
**White box-modellen**
En white box-modell är en modell där vi relativt enklet kan förstå hur modellen kommer fram till sitt resultat.
Exempel:
- Linjär Regression
- Beslutsträd
- Emkel Ridge/Lasso

**Black box-modell**
En black box-modell är svårare att förstå på detaljnivå. Vi kan ge modellen data och få ett resultat, men det kan vara svårt att förklara exakt hur modellen kom fram till resultat.
Exempel:
- Djupa neurala nätverk
- Komplexa ensemble-modeller
- Vissa mycket stora ML-modeller
Vi vet vad som går in och vad som kommer ut, men den interna beslutsprocessen kan vara svår att tolka. 

**Viktig skillnad**
Det är inte alltid helt svart eller vit. En modell kan vara mer eller mindre tolkningsbar.
Till exempel kan en Random Forest vara betydligt svårare att förstp än ett enda beslutsträd, öven om man kan undersöka exempelvis `feature_importances_`

### Uppgift 9. Figuren visar en enkel linjär regressionsmodell, alltså hur man kan beskriva sambandet mellan två variabler med en rät linje.
1. Vad de finns på axlarna?
- x-axeln = Ålder
- y-axeln = Inkomst
Varje svart punkt representerar en person/observation. Punkten visar alltså en persons ålder och motsvarande inkomst.
2. Den blåa linjen är regressionslinjen
Den blå linjen är den skattade regressionslinjen:
\hat y = \hat\theta_0 + \hat\theta_1x
Här betyder:
- \hat y = den förväntade/skattade inkomsten
- \hat\theta_0 = skärningen med y-axeln (intercept)
- \hat\theta_1 = lutningen (slope)
- x = ålder
Så om vi känner till en persons ålder kan vi använda linjen för att uppskatta personen inkomst.
3. Vad betyder \hat\theta_0?
\hat\theta_0 är där den blå linjen skulle skära y-axeln, alltså det värde på inkomsten som modellen förutsäger när

x=0.

I figuren är \hat\theta_0 markerad nära y-axeln.
4. Vad betyder \hat\theta_1?
\hat\theta_1 anger hur mycket den skattade inkomsten förändras när åldern ökar med en enhet.

Eftersom linjen lutar uppåt är \hat\theta_1>0.

Exempelvis, om

\hat\theta_1=500,

skulle modellen säga att inkomsten i genomsnitt ökar med ungefär 500 kr när åldern ökar med ett år.
5. Vad är de röda linjerna?
De röda vertikala linjerna är residualer.

För en observation gäller:

e_i=y_i-\hat y_i

Det betyder:

residual = faktiskt värde − modellens skattade värde

På bilden ser vi exempelvis en punkt ovanför den blå linjen. Då är den faktiska inkomsten högre än den skattade inkomsten:

y_i>\hat y_i

och residualen är positiv.

Om punkten ligger under linjen är:

y_i<\hat y_i

och residualen är negativ

6. Den viktigaste intuitionen

Tänk så här:

Svarta punkter = verkliga observationer
Blå linje = modellens bästa raka uppskattning
Rött avstånd = hur mycket en observation avviker från modellen

Regressionslinjen försöker placera sig så att de totala avvikelserna blir så små som möjligt enligt minsta-kvadratmetoden.

Så figuren illustrerar egentligen hela idén med linjär regression: vi använder en rak linje för att beskriva sambandet mellan ålder och inkomst, och residualerna visar hur bra eller dåligt linjen passar de faktiska observationerna.

### Uppgift 11. Determinationskoefficienten, är ett mått som används för att beskriva hur stor del av variationen i den beroende variabeln som modellen kan förklara. 
**Enkelt exempel**
Om vi försöker förutsäga huspriser och får 
````R2 = 0.80````
kan vi ungefär säga:
Modellen kan förklara 80% av variationen i huspriserna utifrån de variabler som används i modellen.
De resterande 20 % förklaras inte av modellen. 
**Hur värdet ska tolkas?**
Generellt:
````
R² = 1.0   → modellen förklarar all variation
R² = 0.8   → modellen förklarar 80 %
R² = 0.5   → modellen förklarar 50 %
R² = 0     → inte bättre än att alltid använda medelvärdet
````
R2
 ligger ofta mellan 0 och 1, men kan faktiskt bli negativt på testdata om modellen presterar sämre än en modell som alltid förutsäger medelvärdet.

**Viktigt**
**Högre R2 är bättre.**
Det skiljer sig alltså från exempelvis `neg_root_mean_squared_error`, där scikit-learn använder negativa värden eftersom scoring-systemet vill att högre alltid ska vara bättre.

`````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Ladda och undersöka datasetet (EDA)
df = pd.read_csv('diamonds.csv')

# Rensa bort eventuell index-kolumn om den finns
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

print("--- HEAD ---")
print(df.head())
print("\n--- INFO ---")
print(df.info())
print("\n--- SHAPE ---")
print(f"Dimensioner: {df.shape}")
print("\n--- NULL-VÄRDEN ---")
print(df.isnull().sum())
print("\n--- BESKRIVANDE STATISTIK ---")
print(df.describe())

# 2. Dela upp X och y samt Train/Val/Test
X = df.drop(columns=['price'])
y = df['price']

# Först delar vi av testdatan (15%) - används ENDAST i slutet
X_train_val, X_test, y_train_val, y_test = train_test_split(X, y, test_size=0.15, random_state=42)

# Sedan delar vi resten i train (70% av totalen -> ca 82.3% här) och validation (15% av totalen)
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=0.1765, random_state=42)

print("\n--- STORLEKAR PÅ SPLITS ---")
print(f"Train set:       {X_train.shape[0]} rader")
print(f"Validation set:  {X_val.shape[0]} rader")
print(f"Test set:        {X_test.shape[0]} rader")

# 3. Identifiera variabler och prepocessing
numerical_cols = ['carat', 'depth', 'table', 'x', 'y', 'z']
categorical_cols = ['cut', 'color', 'clarity']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ]
)

# 4. Träna och utvärdera modeller på val-set

# --- A. Linear Regression ---
lin_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])
lin_pipeline.fit(X_train, y_train)
lin_val_preds = lin_pipeline.predict(X_val)

lin_mse = mean_squared_error(y_val, lin_val_preds)
lin_rmse = np.sqrt(lin_mse)
lin_mae = mean_absolute_error(y_val, lin_val_preds)
lin_r2 = r2_score(y_val, lin_val_preds)

# --- B. Random Forest (med GridSearchCV) ---
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(random_state=42, n_jobs=-1))
])
rf_param_grid = {
    'model__n_estimators': [50, 100],
    'model__max_depth': [10, 20]
}
rf_grid = GridSearchCV(rf_pipeline, rf_param_grid, cv=3, scoring='neg_root_mean_squared_error', n_jobs=-1)
rf_grid.fit(X_train, y_train)
best_rf = rf_grid.best_estimator_
rf_val_preds = best_rf.predict(X_val)

rf_mse = mean_squared_error(y_val, rf_val_preds)
rf_rmse = np.sqrt(rf_mse)
rf_mae = mean_absolute_error(y_val, rf_val_preds)
rf_r2 = r2_score(y_val, rf_val_preds)

# --- C. SVM (med GridSearchCV) --- 
# Obs: SVM kan vara långsamt på stora dataset, vi begränsar parameterutrymmet något
svm_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", SVR())
])
svm_param_grid = {
    'model__C': [10, 100],
    'model__epsilon': [0.1, 1.0]
}
svm_grid = GridSearchCV(svm_pipeline, svm_param_grid, cv=3, scoring='neg_root_mean_squared_error', n_jobs=-1)
svm_grid.fit(X_train, y_train)
best_svm = svm_grid.best_estimator_
svm_val_preds = best_svm.predict(X_val)

svm_mse = mean_squared_error(y_val, svm_val_preds)
svm_rmse = np.sqrt(svm_mse)
svm_mae = mean_absolute_error(y_val, svm_val_preds)
svm_r2 = r2_score(y_val, svm_val_preds)

# 5. Jämförelse av modeller på valideringsdata
print("\n==============================")
print("       MODELLJÄMFÖRELSE       ")
print("==============================")
print(f"Linear Regression RMSE: {lin_rmse:.2f}")
print(f"Random Forest RMSE:     {rf_rmse:.2f}")
print(f"SVM RMSE:               {svm_rmse:.2f}")
print()
print(f"Linear Regression MAE:  {lin_mae:.2f}")
print(f"Random Forest MAE:      {rf_mae:.2f}")
print(f"SVM MAE:                {svm_mae:.2f}")
print()
print(f"Linear Regression MSE:  {lin_mse:.2f}")
print(f"Random Forest MSE:      {rf_mse:.2f}")
print(f"SVM MSE:                {svm_mse:.2f}")
print()
print(f"Linear Regression R²:   {lin_r2:.3f}")
print(f"Random Forest R²:       {rf_r2:.3f}")
print(f"SVM R²:                 {svm_r2:.3f}")

# 6. Visualiering: Predikterat vc Verkligt pris
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(y_val, lin_val_preds, alpha=0.3, color='blue')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--', lw=2)
plt.xlabel("Verkligt pris")
plt.ylabel("Predikterat pris")
plt.title("Linear Regression")

plt.subplot(1, 3, 2)
plt.scatter(y_val, rf_val_preds, alpha=0.3, color='green')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--', lw=2)
plt.xlabel("Verkligt pris")
plt.ylabel("Predikterat pris")
plt.title("Random Forest")

plt.subplot(1, 3, 3)
plt.scatter(y_val, svm_val_preds, alpha=0.3, color='orange')
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'k--', lw=2)
plt.xlabel("Verkligt pris")
plt.ylabel("Predikterat pris")
plt.title("SVM")

plt.tight_layout()
plt.show()

`````
